# Heart Disease Prediction — Exploratory Analysis

This notebook explores the supplied educational dataset. The modeling workflow is implemented in `train.py`, where preprocessing is fit inside pipelines and evaluated without fitting preprocessing on the held-out test set.

> **Disclaimer:** Not a clinical tool. Do not use for diagnosis or medical decisions.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

ROOT = Path.cwd()
if not (ROOT / 'data' / 'heart.csv').exists():
    ROOT = ROOT.parent
df = pd.read_csv(ROOT / 'data' / 'heart.csv')
df.head()
```

## Dataset overview

```python
print('Shape:', df.shape)
display(df.info())
display(df.describe().T)
```

```python
display(df.isna().sum().rename('missing_values').to_frame())
display(df['target'].value_counts().sort_index().rename('count').to_frame())
```

## Target balance

```python
ax = sns.countplot(data=df, x='target')
ax.set(title='Target class counts', xlabel='Dataset target label', ylabel='Count')
plt.tight_layout()
plt.show()
```

## Numeric feature distributions

```python
df.drop(columns='target').hist(figsize=(15, 12), bins=20)
plt.suptitle('Feature distributions', y=1.02)
plt.tight_layout()
plt.show()
```

## Correlations (exploratory only)

```python
plt.figure(figsize=(12, 9))
sns.heatmap(df.corr(numeric_only=True), cmap='coolwarm', center=0)
plt.title('Feature correlation matrix')
plt.tight_layout()
plt.show()
```
